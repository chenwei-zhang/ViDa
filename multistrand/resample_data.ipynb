```python
import os
import tqdm as tqdm
import numpy as np
import h5py as h5


def read_machinek(inpath, rxn, num_traj):
    def _read_trajectory_h5(fpath, sim_no): 
        import h5py as h5
        with h5.File(fpath, "r") as f:
            times = f[str(sim_no)]["times"][:]
            energies = f[str(sim_no)]["energies"][:]
            structs = [s.decode() for s in f[str(sim_no)]["structs"]]          
            ids = [s.decode() for s in f[str(sim_no)]["ordered_ids"]]
        return times, energies, structs, ids
    
    fpath = os.path.join(inpath, f"{rxn}.hdf5")
    
    trajs_states, trajs_times, trajs_energies, trajs_ids  = [],[],[], []

    for i in tqdm.tqdm(range(num_traj)):
        traj = _read_trajectory_h5(fpath, i)
        trajs_times.append(traj[0])
        trajs_energies.append(traj[1])
        trajs_states.append(traj[2])
        trajs_ids.append(traj[3])
    
    trajs_times = np.array(trajs_times, dtype=object)
    trajs_energies = np.array(trajs_energies, dtype=object)
    trajs_states = np.array(trajs_states, dtype=object)
    trajs_ids = np.array(trajs_ids, dtype=object)
        
    return trajs_states, trajs_times, trajs_energies, trajs_ids


def append_filtered_trajectory(fpath, traj_index, structs, ids, times, energies):
    # Open in append mode
    with h5.File(fpath, "a") as f:
        grp = f.create_group(str(traj_index))
        
        # Convert string lists to byte arrays
        structs = np.array([s.encode("utf-8") for s in structs], dtype="S")
        ids = np.array([s.encode("utf-8") for s in ids], dtype="S")
        
        # Save datasets
        grp.create_dataset("structs", data=structs)
        grp.create_dataset("ordered_ids", data=ids)
        grp.create_dataset("times", data=np.array(times))
        grp.create_dataset("energies", data=np.array(energies))
```

```python
inpath = f'../data/raw_data/machinektest/'
num_traj = 1000
rxn = 'central_toehold8_full'
trajs_states, trajs_times, trajs_energies, trajs_ids = read_machinek(inpath, rxn, num_traj)
```

```python
for no in tqdm.tqdm(range(len(trajs_times))):
    ordered_structs = []
    ordered_ids = []
    times = []
    energies = []   
    timestep = 1e-7
    next_snapshot_time = timestep 

    for i in range(len(trajs_times[no])):
        current_time = trajs_times[no][i]
        
        if_record = (i == 0) or (i == len(trajs_times[no])-1) or (current_time >= next_snapshot_time)
        
        if if_record:
            ordered_structs.append(trajs_states[no][i])
            ordered_ids.append(trajs_ids[no][i])
            times.append(trajs_times[no][i])
            energies.append(trajs_energies[no][i])
            
            while next_snapshot_time <= current_time:
                next_snapshot_time += timestep
    
    # save to hdf5
    traj_filename = "../data/raw_data/machinektest/central_toehold8.hdf5"
    append_filtered_trajectory(traj_filename, no, ordered_structs, ordered_ids, times, energies)

    print(f"Saved simulation {no}")  
    
print("All simulations saved.")
```
